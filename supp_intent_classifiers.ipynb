```python
import spacy
from spacy.training.example import Example

nlp_wine = spacy.blank("en")
textcat = nlp_wine.add_pipe("textcat")

# Add labels
textcat.add_label("similar")
textcat.add_label("pairing")
textcat.add_label("description")

optimizer = nlp_wine.begin_training()

for i in range(30):
    losses = {}
    for text, annotation in WINE_TRAIN_DATA:
        example = Example.from_dict(nlp_wine.make_doc(text), annotation)
        nlp_wine.update([example], sgd=optimizer, losses=losses)
    print(f"Epoch {i+1} Losses: {losses}")

nlp_wine.to_disk("wine_intent_model")
```
